```python
import pandas as pd

df = pd.read_csv('Iris.csv')
df.head()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

plt.figure(figsize=(8,5))

for species in df['Species'].unique():
    subset = df[df['Species'] == species]
    plt.hist(subset['PetalLengthCm'],
             bins=10,
             alpha=0.5,
             label=species)

plt.title('Petal Length Distribution by Species')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Frequency')
plt.legend()
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(8,5))

sns.scatterplot(
    x='PetalLengthCm',
    y='PetalWidthCm',
    hue='Species',
    data=df,
    s=80
)

plt.title('Petal Length vs Petal Width')
plt.xlabel('Petal Length (cm)')
plt.ylabel('Petal Width (cm)')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


# Select only numerical columns
corr_matrix = df[['SepalLengthCm',
                  'SepalWidthCm',
                  'PetalLengthCm',
                  'PetalWidthCm']].corr()

plt.figure(figsize=(8,6))

sns.heatmap(
    corr_matrix,
    annot=True,
    cmap='coolwarm',
    fmt='.2f'
)

plt.title('Correlation Heatmap of Iris Features')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns


plt.figure(figsize=(8,5))

sns.violinplot(
    x='Species',
    y='PetalLengthCm',
    data=df
)

plt.title('Petal Length Distribution by Species')
plt.xlabel('Species')
plt.ylabel('Petal Length (cm)')
plt.show()
```
