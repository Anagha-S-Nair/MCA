```python
import pandas as pd

df = pd.read_csv('Iris.csv')
df.head()
```

```python
if "Id" in df.columns:
    df = df.drop("Id", axis=1)

print(df.head())
```

```python
X = df.iloc[:, :-1].values
y = df.iloc[:, -1].values

print("Features:")
print(X[:5])

print("\nLabels:")
print(y[:5])
```

```python
split = int(0.8 * len(df))

X_train = X[:split]
y_train = y[:split]

X_test = X[split:]
y_test = y[split:]

print("Training data:", len(X_train))
print("Testing data:", len(X_test))
```

```python
import math

def euclidean_distance(point1, point2):

    distance = 0

    for i in range(len(point1)):
        distance += (point1[i] - point2[i]) ** 2

    return math.sqrt(distance)
```

```python
point1 = [5.1, 3.5, 1.4, 0.2]
point2 = [4.9, 3.0, 1.4, 0.2]

distance = euclidean_distance(point1, point2)

print("Distance:", distance)
```

```python
point1 = [5.1, 3.5, 1.4, 0.2]
point2 = [4.9, 3.0, 1.4, 0.2]

distance = euclidean_distance(point1, point2)

print("Distance:", distance)
```

```python
def find_neighbors(X_train, y_train, new_instance, k):

    distances = []

    for i in range(len(X_train)):

        distance = euclidean_distance(
            new_instance,
            X_train[i]
        )

        distances.append((distance, y_train[i]))

    # Sort according to distance
    distances.sort(key=lambda x: x[0])

    # Select K nearest neighbors
    neighbors = distances[:k]

    return neighbors
```

```python
new_instance = [5.1, 3.5, 1.4, 0.2]

k = 5

neighbors = find_neighbors(
    X_train,
    y_train,
    new_instance,
    k
)

print("Nearest Neighbors:")

for distance, label in neighbors:
    print("Distance:", distance, "Label:", label)
```

```python
from collections import Counter

def majority_vote(neighbors):

    labels = []

    for distance, label in neighbors:
        labels.append(label)

    counter = Counter(labels)

    prediction = counter.most_common(1)[0][0]

    return prediction
```

```python
new_instance = [5.1, 3.5, 1.4, 0.2]

k = 5

neighbors = find_neighbors(
    X_train,
    y_train,
    new_instance,
    k
)

prediction = majority_vote(neighbors)

print("New Instance:", new_instance)
print("Predicted Label:", prediction)
```

```python
correct = 0

for i in range(len(X_test)):

    neighbors = find_neighbors(
        X_train,
        y_train,
        X_test[i],
        k
    )

    prediction = majority_vote(neighbors)

    print(
        "Actual:",
        y_test[i],
        "Predicted:",
        prediction
    )

    if prediction == y_test[i]:
        correct += 1
```

```python
accuracy = (correct / len(X_test)) * 100

print("\nAccuracy:", accuracy, "%")
```
